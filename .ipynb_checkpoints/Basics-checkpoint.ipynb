# Control Flow

```python
a = int(input("Enter nu"))
b = int(input("Enter sec no"))
if a > b:
    print(a,"A is greater")
else:
    print(b,"B is Greater")
```

```python
a = float(input("Enter first no: "))
b = float(input("Enter second no: "))
if b>a:
    print("b is greater")
elif a == b:
    print("Both are equal")
else:
    print("a is greater")
```

```python
#Nested if
x = int(input("enter a number: "))
if x > 10:
    print("Your number is above 10")
    print("Inside top if")
    if x > 20:
        print("Your number is Above 20")
        print("Inside nested if")
    else:
        print("Smaller or Equal to 20")
        print("Inside the else part of nested if")
print("Outside all ifs")
```

```python
x = float(input("Enter a real no: "))
y = round(x) 
if x > 0:
    if y > x:
        intProtion = y-1   #29.6 
    else:
        intProtion = y
else:
    if y < x:
        intProtion = y+1
    else:
        intProtion = y
    
if intProtion % 2 == 0:
    print("Even")
else:
    print("Odd")
print(intProtion)
```

# Loops

```python
n = int(input())
i = 1
while i < n:
    print(i**2)
    print("This is iteration number:",i)
    i += 1  #i = i+1
print("Done")
```

```python
n = int(input("Max Iterations: "))
i = 1
while i < n:
    if i%2 == 0:
        print(i)
    else:
        pass
    i = i+1
print("Done")
```

```python
i = 10
while True:
    if i%17 == 0:
        print("Break")
        break
    else:
        i += 1
        continue
    print("I am Inside the loop")
print("Done")
```
